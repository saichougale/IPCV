```python
#displayng the image
import cv2
from google.colab.patches import cv2_imshow
image_path = "/content/drive/My Drive/IPCV/image.jpg"
image = cv2.imread(image_path)
cv2_imshow(image)
```

```python
# Extracting the height and width of an image
h, w = image.shape[:2]
# Displaying the height and width
print("Height = {}, Width = {}".format(h, w))
```

```python
#resizing the image
resize = cv2.resize(image, (400, 600))
cv2_imshow(resize)
cv2.waitKey(0)
```

```python
#displaying text
output = image.copy()
text=cv2.putText(output, "Welcome to Image Processing", (300, 250), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
cv2_imshow(text)
```

```python
#displaying the region of image
# We will calculate the region of interest
# by slicing the pixels of the image
roi = image[105 : 150, 20 : 70]
cv2_imshow(roi)
```

```python
#drawing the rectangle
output = image.copy()
rectangle = cv2.rectangle(output, (500, 500),
                        (200, 200), (255, 0, 0), 2)
cv2_imshow(rectangle)
```

```python
#Rotating image that we resized previously by 270 degree
rotated1 = cv2.rotate(resize, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated1)
```

```python
#Rotataing image by 90 degree clckwise
rotated2 = cv2.rotate(resize, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated2)
```

```python
#color spaces
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

image_lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

cv2_imshow(image)
cv2_imshow(image_rgb)
cv2_imshow(image_hsv)
cv2_imshow(image_gray)
cv2_imshow(image_lab)
```
